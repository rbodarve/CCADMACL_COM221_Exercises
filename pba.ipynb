```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL of the page to scrape
url = "https://basketball.realgm.com/international/league/131/PBA--Commissioners-Cup/stats/2024/Averages/All/All/points/All/desc/1/Regular_Season"

# Send a GET request to fetch the page content
response = requests.get(url)
response.raise_for_status()  # Raise an error if the request fails

# Parse the page content with BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Find the table on the webpage (the table containing the stats)
table = soup.find('table', {'class': 'tablesaw'})

# Extract headers (column names)
headers = [header.get_text(strip=True) for header in table.find_all('th')]

# Extract rows of data
rows = []
for row in table.find_all('tr')[1:]:  # Skip the header row
    cells = row.find_all('td')
    row_data = [cell.get_text(strip=True) for cell in cells]
    if row_data:  # Avoid empty rows
        rows.append(row_data)

# Create a DataFrame to store the data
df = pd.DataFrame(rows, columns=headers)

# Print the DataFrame to see the results
print(df)

# Optionally, save the DataFrame to a CSV file
df.to_csv('pba_commissioners_cup_2024_stats.csv', index=False)
```
